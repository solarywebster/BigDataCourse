```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan,
              50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
               'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})

products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})

print("Таблица sales:")
print(sales)

print("\nТаблица products:")
print(products)
```

Задание 1: Найти количество пропусков в каждом столбце

```python
missing_values = sales.isnull().sum()

print("Количество пропусков в каждом столбце:")
print(missing_values)
```

Задание 2: Найти среднюю цену и количество для каждого продукта

```python
product_price_stats = sales.groupby('Product')['Price'].agg(['mean', 'count'])

print("Средняя цена и количество непропущенных значений Price по товарам:")
print(product_price_stats)
```

Задание 3: Удалить строки без цены или региона

```python
sales_without_nan = sales.dropna(subset=['Price', 'Region'])

print("Таблица после удаления строк с NaN в Price или Region:")
print(sales_without_nan)
```

Задание 4: Переименовать все столбцы на верхний регистр.

```python
sales_upper = sales.copy()
sales_upper.columns = sales_upper.columns.str.upper()

print("Таблица с названиями столбцов в верхнем регистре:")
print(sales_upper)

print("\nНазвания столбцов:")
print(sales_upper.columns)
```

Задание 5: Горизонтально объединить sales (первые 3 строки) и products (первые 3 строки)

```python
horizontal_concat = pd.concat(
    [
        sales.head(3).reset_index(drop=True),
        products.head(3).reset_index(drop=True)
    ],
    axis=1
)

print("Горизонтальное объединение первых 3 строк sales и products:")
print(horizontal_concat)
```
