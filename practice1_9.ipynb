Импорт и создание DataFrame:

```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)

print("Исходный DataFrame:")
print(df)
```

Задание 1: Количество строк и столбцов

```python
print("Количество строк и столбцов:", df.shape)
```

Задание 2: строки с нечётными индексами

```python
odd_index_rows = df.iloc[1::2]

print("Строки с нечётными индексами:")
print(odd_index_rows)
```

Задание 3: Выбрать строки, где Salary выше медианы.

```python
median_salary = df['Salary'].median()
print("\n3. Медиана Salary:", median_salary)
print(df[df['Salary'] > median_salary])
```

Задание 4: Количество людей из каждого города

```python
print("\n4. Люди по городам:")
print(df['City'].value_counts())
```

Задание 5: Создать столбец City_Code

```python
df['City_Code'] = df['City'].map({'Moscow': 'MSK', 'SPb': 'SPB', 'Kazan': 'KZN'})
print("\n5. Таблица с City_Code:")
print(df)
```
